## 12.5.1 Principal Components Analysis

We perform PCA on `USArrests` — 50 US states (rows) and four variables (columns), fetched from R via `get_rdataset()`.

```python
USArrests = get_rdataset('USArrests').data
USArrests.head()
```

```python
USArrests.columns
```

The variables have **very different means and variances**, so PCA on the raw data would be dominated by `Assault`.

```python
USArrests.mean()
```

```python
USArrests.var()
```

So we **standardize** each variable to mean 0 and standard deviation 1 with `StandardScaler`, then fit `PCA()`.

```python
scaler = StandardScaler(with_std=True, with_mean=True)
USArrests_scaled = scaler.fit_transform(USArrests)

pcaUS = PCA()
pcaUS.fit(USArrests_scaled)
pcaUS.mean_          # all ~0 because we centered the data
```

```python
scores = pcaUS.transform(USArrests_scaled)
```

Each row of `components_` is a principal-component **loading vector**.

```python
pcaUS.components_
```

### Biplot
`sklearn` has no built-in biplot, so we draw one. We reproduce the book's Figure 12.1 orientation by flipping the sign of PC2 (components are only defined up to sign) and lengthening the arrows.

```python
i, j = 0, 1                # which components
scale_arrow = s_ = 2
scores[:, 1] *= -1
pcaUS.components_[1] *= -1  # flip the y-axis

fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(scores[:, 0], scores[:, 1])
ax.set_xlabel('PC%d' % (i + 1))
ax.set_ylabel('PC%d' % (j + 1))
for k in range(pcaUS.components_.shape[1]):
    ax.arrow(0, 0, s_ * pcaUS.components_[i, k], s_ * pcaUS.components_[j, k],
             color='red', head_width=0.04)
    ax.text(s_ * pcaUS.components_[i, k], s_ * pcaUS.components_[j, k],
            USArrests.columns[k], color='red', fontsize=13);
```

### Variance explained

```python
scores.std(0, ddof=1)
```

```python
pcaUS.explained_variance_
```

```python
pcaUS.explained_variance_ratio_   # PVE
```

The first component explains **62.0%** of the variance, the second **24.7%**, and so on. We plot the PVE and the cumulative PVE (a scree plot).

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ticks = np.arange(pcaUS.n_components_) + 1
ax = axes[0]
ax.plot(ticks, pcaUS.explained_variance_ratio_, marker='o')
ax.set_xlabel('Principal Component')
ax.set_ylabel('Proportion of Variance Explained')
ax.set_ylim([0, 1]); ax.set_xticks(ticks)
ax = axes[1]
ax.plot(ticks, pcaUS.explained_variance_ratio_.cumsum(), marker='o')
ax.set_xlabel('Principal Component')
ax.set_ylabel('Cumulative Proportion of Variance Explained')
ax.set_ylim([0, 1]); ax.set_xticks(ticks);
```

## 12.5.2 Matrix Completion

Solving the low-rank optimization problem on a centered matrix is equivalent to computing its principal components, and the **SVD** solves it. We use the scaled `USArrests` matrix as `X`.

```python
X = USArrests_scaled
U, D, V = np.linalg.svd(X, full_matrices=False)
U.shape, D.shape, V.shape
```

```python
V                       # equals the loading matrix up to a sign flip
```

```python
pcaUS.components_
```

Multiplying each column of `U` by the matching element of `D` recovers the PCA scores (up to sign).

```python
(U * D[None, :])[:3]
```

### Omitting 20 entries at random
We drop one entry in each of 20 randomly chosen states, so every row keeps at least three values.

```python
n_omit = 20
np.random.seed(15)
r_idx = np.random.choice(np.arange(X.shape[0]), n_omit, replace=False)
c_idx = np.random.choice(np.arange(X.shape[1]), n_omit, replace=True)
Xna = X.copy()
Xna[r_idx, c_idx] = np.nan
```

A rank-`M` approximation via the SVD (Step 2 of Algorithm 12.1):

```python
def low_rank(X, M=1):
    U, D, V = np.linalg.svd(X)
    L = U[:, :M] * D[None, :M]
    return L.dot(V[:M])
```

**Step 1** — initialize missing entries with column means:

```python
Xhat = Xna.copy()
Xbar = np.nanmean(Xhat, axis=0)
Xhat[r_idx, c_idx] = Xbar[c_idx]
```

```python
thresh = 1e-7
rel_err = 1
count = 0
ismiss = np.isnan(Xna)
mssold = np.mean(Xhat[~ismiss] ** 2)
mss0 = np.mean(Xna[~ismiss] ** 2)
```

**Step 2** — iterate: low-rank approximation → refill missing → measure relative error, until it drops below `1e-7`.

```python
while rel_err > thresh:
    count += 1
    Xapp = low_rank(Xhat, M=1)          # 2(a)
    Xhat[ismiss] = Xapp[ismiss]         # 2(b)
    mss = np.mean(((Xna - Xapp)[~ismiss]) ** 2)  # 2(c)
    rel_err = (mssold - mss) / mss0
    mssold = mss
    print("Iteration: {0}, MSS:{1:.3f}, Rel.Err {2:.2e}".format(count, mss, rel_err))
```

Correlation between the 20 imputed values and the true values:

```python
np.corrcoef(Xapp[ismiss], X[ismiss])[0, 1]
```

## 12.5.3 Clustering

### K-Means
A simulated example with two genuine clusters: the first 25 points are shifted relative to the rest.

```python
np.random.seed(0)
X = np.random.standard_normal((50, 2))
X[:25, 0] += 3
X[:25, 1] -= 4
```

```python
kmeans = KMeans(n_clusters=2, random_state=2, n_init=20).fit(X)
kmeans.labels_
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(X[:, 0], X[:, 1], c=kmeans.labels_)
ax.set_title("K-Means Clustering Results with K=2");
```

With `K=3`, K-means splits the data further:

```python
kmeans = KMeans(n_clusters=3, random_state=3, n_init=20).fit(X)
fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(X[:, 0], X[:, 1], c=kmeans.labels_)
ax.set_title("K-Means Clustering Results with K=3");
```

`n_init` runs multiple random starts and keeps the best (lowest within-cluster sum of squares). Always use a large value to avoid poor local optima.

```python
kmeans1  = KMeans(n_clusters=3, random_state=3, n_init=1).fit(X)
kmeans20 = KMeans(n_clusters=3, random_state=3, n_init=20).fit(X)
kmeans1.inertia_, kmeans20.inertia_
```

### Hierarchical Clustering
`AgglomerativeClustering` (abbreviated `HClust`) with complete, average and single linkage.

```python
HClust = AgglomerativeClustering
hc_comp = HClust(distance_threshold=0, n_clusters=None, linkage='complete')
hc_comp.fit(X)
```

```python
hc_avg = HClust(distance_threshold=0, n_clusters=None, linkage='average')
hc_avg.fit(X)
hc_sing = HClust(distance_threshold=0, n_clusters=None, linkage='single')
hc_sing.fit(X);
```

The same single-linkage result via a precomputed distance matrix:

```python
D = np.zeros((X.shape[0], X.shape[0]))
for i in range(X.shape[0]):
    x_ = np.multiply.outer(np.ones(X.shape[0]), X[i])
    D[i] = np.sqrt(np.sum((X - x_) ** 2, 1))
hc_sing_pre = HClust(distance_threshold=0, n_clusters=None,
                     metric='precomputed', linkage='single')
hc_sing_pre.fit(D);
```

`compute_linkage` (from `ISLP.cluster`) converts the fitted model into a SciPy linkage matrix that `dendrogram` can plot.

```python
cargs = {'color_threshold': -np.inf, 'above_threshold_color': 'black'}
linkage_comp = compute_linkage(hc_comp)
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_comp, ax=ax, **cargs);
```

Coloring branches that merge below height 4:

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_comp, ax=ax, color_threshold=4, above_threshold_color='black');
```

Cluster labels from cutting the tree — by number of clusters, or by height:

```python
cut_tree(linkage_comp, n_clusters=4).T
```

```python
cut_tree(linkage_comp, height=5).T
```

Scaling the features before clustering:

```python
scaler = StandardScaler()
X_scale = scaler.fit_transform(X)
hc_comp_scale = HClust(distance_threshold=0, n_clusters=None,
                       linkage='complete').fit(X_scale)
linkage_comp_scale = compute_linkage(hc_comp_scale)
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_comp_scale, ax=ax, **cargs)
ax.set_title("Hierarchical Clustering with Scaled Features");
```

Correlation-based dissimilarity (needs ≥3 features), using `1 - corr` as the distance:

```python
X = np.random.standard_normal((30, 3))
corD = 1 - np.corrcoef(X)
hc_cor = HClust(linkage='complete', distance_threshold=0,
                n_clusters=None, metric='precomputed')
hc_cor.fit(corD)
linkage_cor = compute_linkage(hc_cor)
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_cor, ax=ax, **cargs)
ax.set_title("Complete Linkage with Correlation-Based Dissimilarity");
```

## 12.5.4 NCI60 Data Example
6,830 gene-expression measurements on 64 cancer cell lines. The cancer-type labels are held out and used only to check the unsupervised results afterward.

```python
NCI60 = load_data('NCI60')
nci_labs = NCI60['labels']
nci_data = NCI60['data']
nci_data.shape
```

```python
nci_labs.value_counts()
```

### PCA on NCI60 (scaled genes)

```python
scaler = StandardScaler()
nci_scaled = scaler.fit_transform(nci_data)
nci_pca = PCA()
nci_scores = nci_pca.fit_transform(nci_scaled)
```

```python
cancer_types = list(np.unique(nci_labs))
nci_groups = np.array([cancer_types.index(lab) for lab in nci_labs.values])
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ax = axes[0]
ax.scatter(nci_scores[:, 0], nci_scores[:, 1], c=nci_groups, marker='o', s=50)
ax.set_xlabel('PC1'); ax.set_ylabel('PC2')
ax = axes[1]
ax.scatter(nci_scores[:, 0], nci_scores[:, 2], c=nci_groups, marker='o', s=50)
ax.set_xlabel('PC1'); ax.set_ylabel('PC3');
```

Cell lines of the same cancer type tend to sit near each other. Scree / cumulative-PVE plots:

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ticks = np.arange(nci_pca.n_components_) + 1
ax = axes[0]
ax.plot(ticks, nci_pca.explained_variance_ratio_, marker='o')
ax.set_xlabel('Principal Component'); ax.set_ylabel('PVE')
ax = axes[1]
ax.plot(ticks, nci_pca.explained_variance_ratio_.cumsum(), marker='o')
ax.set_xlabel('Principal Component'); ax.set_ylabel('Cumulative PVE');
```

The first ~7 components explain about 40% of the variance, with an elbow after the seventh.

### Hierarchical clustering of the cell lines

```python
def plot_nci(linkage, ax, cut=-np.inf):
    cargs = {'above_threshold_color': 'black', 'color_threshold': cut}
    hc = HClust(n_clusters=None, distance_threshold=0,
                linkage=linkage.lower()).fit(nci_scaled)
    linkage_ = compute_linkage(hc)
    dendrogram(linkage_, ax=ax, labels=np.asarray(nci_labs),
               leaf_font_size=10, **cargs)
    ax.set_title('%s Linkage' % linkage)
    return hc
```

```python
fig, axes = plt.subplots(3, 1, figsize=(15, 30))
hc_comp = plot_nci('Complete', axes[0])
hc_avg  = plot_nci('Average',  axes[1])
hc_sing = plot_nci('Single',   axes[2]);
```

Single linkage produces trailing clusters; complete and average give more balanced ones. Cut the complete tree into four clusters and cross-tabulate against cancer type:

```python
linkage_comp = compute_linkage(hc_comp)
comp_cut = cut_tree(linkage_comp, n_clusters=4).reshape(-1)
pd.crosstab(nci_labs['label'], pd.Series(comp_cut, name='Complete'))
```

All leukemia lines land in one cluster; breast cancer spreads across three. Draw the height-140 cut that yields these four clusters:

```python
fig, ax = plt.subplots(figsize=(10, 10))
plot_nci('Complete', ax, cut=140)
ax.axhline(140, c='r', linewidth=4);
```

### Hierarchical vs K-means (K=4)

```python
nci_kmeans = KMeans(n_clusters=4, random_state=0, n_init=20).fit(nci_scaled)
pd.crosstab(pd.Series(comp_cut, name='HClust'),
            pd.Series(nci_kmeans.labels_, name='K-means'))
```

Cluster identifiers are arbitrary, but the two partitions clearly differ. Finally, cluster on the **first five PC score vectors** (a denoised version of the data):

```python
hc_pca = HClust(n_clusters=None, distance_threshold=0,
                linkage='complete').fit(nci_scores[:, :5])
linkage_pca = compute_linkage(hc_pca)
fig, ax = plt.subplots(figsize=(10, 8))
dendrogram(linkage_pca, labels=np.asarray(nci_labs), leaf_font_size=10, ax=ax, **cargs)
ax.set_title("Hier. Clust. on First Five Score Vectors")
pca_labels = pd.Series(cut_tree(linkage_pca, n_clusters=4).reshape(-1), name='Complete-PCA')
pd.crosstab(nci_labs['label'], pca_labels)
```
