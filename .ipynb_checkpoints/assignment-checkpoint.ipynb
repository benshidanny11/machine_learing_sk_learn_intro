```python
import pandas as pd
import numpy as np

data = {
    'Hours_Studied': [2, 4, 6, 8, 10],
    'Sleep_Hours': [9, 8, 7, 6, 5],
    'Pass': [0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)
X = df[['Hours_Studied', 'Sleep_Hours']].values
y = df['Pass'].values

# Step 2: Split into training and testing manually (80%-20%)
X_train, X_test = X[:4], X[4:]
y_train, y_test = y[:4], y[4:]


def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def train_logistic_regression(X, y, lr=0.1, epochs=1000):
    n_samples, n_features = X.shape
    weights = np.zeros(n_features)
    bias = 0

    for _ in range(epochs):
        linear_model = np.dot(X, weights) + bias
        y_predicted = sigmoid(linear_model)

        # Gradients
        dw = (1/n_samples) * np.dot(X.T, (y_predicted - y))
        db = (1/n_samples) * np.sum(y_predicted - y)

        # Update weights
        weights -= lr * dw
        bias -= lr * db

    return weights, bias

def predict_logistic(X, weights, bias):
    linear_model = np.dot(X, weights) + bias
    y_predicted = sigmoid(linear_model)
    return np.where(y_predicted >= 0.5, 1, 0)

w_log, b_log = train_logistic_regression(X_train, y_train)
y_pred_log = predict_logistic(X_test, w_log, b_log)
acc_log = np.mean(y_pred_log == y_test)

# ------------------------
# Simple SVM (Linear, using hinge loss)
# ------------------------
def train_svm(X, y, lr=0.01, epochs=1000, C=1):
    n_samples, n_features = X.shape
    y_ = np.where(y == 0, -1, 1)  # Convert to -1,1
    weights = np.zeros(n_features)
    bias = 0

    for _ in range(epochs):
        for idx, x_i in enumerate(X):
            condition = y_[idx] * (np.dot(x_i, weights) + bias) >= 1
            if condition:
                weights -= lr * (2 * 1/epochs * weights)
            else:
                weights -= lr * (2 * 1/epochs * weights - np.dot(x_i, y_[idx]))
                bias -= lr * y_[idx]
    return weights, bias

def predict_svm(X, weights, bias):
    linear_output = np.dot(X, weights) + bias
    return np.where(linear_output >= 0, 1, 0)

w_svm, b_svm = train_svm(X_train, y_train)
y_pred_svm = predict_svm(X_test, w_svm, b_svm)
acc_svm = np.mean(y_pred_svm == y_test)

def random_forest_predict(X):
    preds = []
    for x in X:
        # Rule 1: If Hours_Studied >= 6 -> Pass
        if x[0] >= 6:
            preds.append(1)
        else:
            preds.append(0)
    return np.array(preds)

y_pred_rf = random_forest_predict(X_test)
acc_rf = np.mean(y_pred_rf == y_test)


print("Accuracy Scores (using only Pandas & NumPy):")
print(f"Logistic Regression: {acc_log:.2f}")
print(f"SVM: {acc_svm:.2f}")
print(f"Random Forest: {acc_rf:.2f}")

best_model = max({'Logistic Regression': acc_log, 'SVM': acc_svm, 'Random Forest': acc_rf}, key=lambda k: {'Logistic Regression': acc_log, 'SVM': acc_svm, 'Random Forest': acc_rf}[k])
print(f"\nBest model: {best_model}")
```

```python
import pandas as pd
import numpy as np

# Step 1: Create the dataset
data = {
    'X1': [1, 2, 3, 6, 7],
    'X2': [2, 3, 3, 5, 8],
    'Y':  [0, 0, 1, 1, 1]
}
df = pd.DataFrame(data)

# Features and labels
X_train = df[['X1', 'X2']].values
y_train = df['Y'].values

# Step 2: Define KNN prediction function
def knn_predict(X_train, y_train, x_test, k=3):
    # Compute Euclidean distances
    distances = np.sqrt(np.sum((X_train - x_test) ** 2, axis=1))
    
    # Get indices of k nearest neighbors
    k_indices = np.argsort(distances)[:k]
    
    # Get labels of k nearest neighbors
    k_labels = y_train[k_indices]
    
    # Return majority vote
    counts = np.bincount(k_labels)
    return np.argmax(counts)

# Step 3: Predict for point (4,4) with K=5
x_test = np.array([4, 4])
k = 5
predicted_label = knn_predict(X_train, y_train, x_test, k)

print(f"Predicted label for point {x_test} with K={k}: {predicted_label}")
```
