# Predict Food Waste to optimise income

```python
# ===========================
# Importing libraries
# ===========================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

#### Load dataset

```python
# Source: https://www.kaggle.com/datasets/trevinhannibal/food-wastage-data-in-restaurant/data
# ---------------------------
file_path = "food_wastage_data.csv"
df = pd.read_csv(file_path)
```

#### Clean column names to snake_case

```python

def to_snake_case(s):
    return s.strip().lower().replace(' ', '_').replace('-', '_').replace('/', '_')
df.columns = [to_snake_case(c) for c in df.columns]
```

#### Data pre processing

```python
## CHeck missing values
print("Missing values before cleaning:\n", df.isnull().sum()) 

# Handle Outliers (IQR Method)
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]

# Remove outliers for both columns
df = remove_outliers_iqr(df, "number_of_guests")
df = remove_outliers_iqr(df, "quantity_of_food")
```

#### Define features and target

```python
target_col = "wastage_food_amount"
y = df[target_col]
X = df.drop(columns=[target_col])
```

#### Handle categorical variables

```python
cat_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True)
```

#### Train/test split

```python

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)
```

```python
# ---------------------------
# Train model
# ---------------------------
model = DecisionTreeRegressor(max_depth=9, random_state=42)
model.fit(X_train, y_train)
```

```python
# ---------------------------
# 7. Evaluate model
# ---------------------------
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Evaluation Results:")
print(f" MAE  : {mae:.2f}")
print(f" MSE  : {mse:.2f}")
print(f" RMSE : {rmse:.2f}")
print(f" R²   : {r2:.2f}")
```

```python
# ---------------------------
# 8. Compare predictions vs actuals
# ---------------------------
compare = pd.DataFrame({"Actual": y_test.values, "Predicted": np.round(y_pred, 2)})
print("\nSample Predictions vs Actuals:")
print(compare.head(10))
```

```python
# ---------------------------
# 9. Feature importance
# ---------------------------
feat_imp = pd.Series(model.feature_importances_, index=X_encoded.columns)
feat_imp = feat_imp.sort_values(ascending=False)

print("\nTop 10 Important Features:")
print(feat_imp.head(10))
```

```python
# ---------------------------
# 10. Visualize decision tree
# ---------------------------
plt.figure(figsize=(30,15))
plot_tree(model, feature_names=X_encoded.columns, filled=True, rounded=True, fontsize=8)
plt.title("Decision Tree (max_depth=9)")
plt.show()
```

```python
#Parameter tuning:
param_grid = {
    "criterion": ["squared_error", "friedman_mse", "absolute_error"],
    "max_depth": range(2,20)
}

# Setup GridSearchCV
grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5)

# Fit the model
grid.fit(X_train, y_train)

# Best parameters
print("Best parameters:", grid.best_params_)
```
