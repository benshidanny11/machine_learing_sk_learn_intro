## Ploting Data from data frame

```python
car_sales=pd.read_csv('car-sales.csv')
car_sales
```

```python
car_sales['Price'] = car_sales['Price'].str.replace(r'[\$,.]', '', regex=True)
car_sales
```

```python
car_sales['Price']=car_sales['Price'].str[:-2]
```

```python
car_sales
```

```python
car_sales['Sale Date']=pd.date_range('1/1/2020', periods=len(car_sales))
```

```python
car_sales
```

```python
car_sales['Total sales']=car_sales['Price'].astype(int).cumsum()
car_sales
```

```python
car_sales.plot(x="Sale Date", y="Total sales")
```

```python
car_sales['Price']=car_sales['Price'].astype(int)
car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter")

```

```python
car_sales["Odometer (KM)"].plot.hist()
```

## Using scatter plot on dataframe

```python
over_50=heart_diseases[heart_diseases["age"]>50]
```

```python
over_50.plot.hist()
```

## Scatter on Age and cholestorol with coloring target

```python
over_50.plot(kind='scatter',x='age',  y='chol', c='target');
```

```python
## Using OO method mixing with pyplot method
fig, ax=plt.subplots(figsize=(10, 6))

over_50.plot(kind='scatter',x='age',  y='chol', c='target', ax=ax);
```

```python
## Using OO method from scratch
fig, ax=plt.subplots(figsize=(10, 6))

scatter=ax.scatter(x=over_50["age"], y=over_50["chol"],c=over_50['target'])

ax.set(title="Heart Disease and colesterol levels", xlabel="Ages", ylabel="Chol");

# Add legend

ax.legend(*scatter.legend_elements(), title="Target")

ax.axhline(over_50["chol"].mean(), linestyle="--")
```

```python
## Sub plot of chol and thalach

fig, (ax0, ax1)=plt.subplots(nrows=2, ncols=1,figsize=(10,6))

scatter=ax0.scatter(x=over_50["age"], y=over_50["chol"],c=over_50['target'], cmap="winter")

ax0.set(title="Heart Disease and colesterol levels", xlabel="Ages", ylabel="Chol");

# Add legend

ax0.legend(*scatter.legend_elements(), title="Target")

ax0.axhline(over_50["chol"].mean(), linestyle="--")

#-----------------------------------

scatter=ax1.scatter(x=over_50["age"], y=over_50["thalach"],c=over_50['target'], cmap="winter")

ax1.set(title="Heart Disease and max heard rate levels", xlabel="Ages", ylabel="thalach");

# Add legend

ax1.legend(*scatter.legend_elements(), title="Target")

ax1.axhline(over_50["thalach"].mean(), linestyle="--")

fig.suptitle("Heart disease analysis", fontsize=16, fontweight="bold")
```
